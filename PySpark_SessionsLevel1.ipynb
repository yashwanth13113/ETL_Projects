```python
print("\n--- RDD Transformations ---")

# 1. map(): Applies a function to each element of the RDD
mapped_rdd = num_rdd.map(lambda x: x * 2)
print("Mapped RDD (num_rdd * 2):", mapped_rdd.collect())

# 2. filter(): Selects elements that satisfy a function
filtered_rdd = num_rdd.filter(lambda x: x > 5)
print("Filtered RDD (num_rdd > 5):", filtered_rdd.collect())

# 3. flatMap(): Similar to map, but each input item can be mapped to 0 or more output items
# Split each sentence into words and create a single RDD of words
words_flat_rdd = word_rdd.flatMap(lambda line: line.split(" "))
print("flatMap RDD (words from word_rdd):", words_flat_rdd.collect())

# 4. union(): Returns a new RDD containing all elements from the source RDDs
union_rdd = num_rdd.union(sc.parallelize([100, 200]))
print("Union RDD (num_rdd + [100, 200]):", union_rdd.collect())

# 5. intersection(): Returns a new RDD containing only the elements that are present in both RDDs
intersection_rdd = num_rdd.intersection(sc.parallelize([2, 5, 88]))
print("Intersection RDD (num_rdd and [2, 5, 88]):", intersection_rdd.collect())

# 6. distinct(): Returns a new RDD containing the distinct elements of the source RDD.
distinct_rdd = num_rdd.distinct()
print("Distinct RDD (num_rdd):", distinct_rdd.collect())

# 7. groupByKey() (for Pair RDDs): Groups the values for each key in the RDD.
# This operation is generally expensive as it involves shuffling.
grouped_pair_rdd = pair_rdd.groupByKey().mapValues(list)
print("Grouped Pair RDD (pair_rdd groupByKey):", grouped_pair_rdd.collect())

# 8. reduceByKey() (for Pair RDDs): Combines values for each key using an associative and commutative function.
# More efficient for aggregation than groupByKey followed by map.
reduced_pair_rdd = pair_rdd.reduceByKey(lambda a, b: a + b)
print("Reduced Pair RDD (pair_rdd reduceByKey):", reduced_pair_rdd.collect())
```

### Introduction to PySpark DataFrames

**What is a DataFrame?**

In PySpark, a DataFrame is a distributed collection of data organized into named columns. It's conceptually equivalent to a table in a relational database or a DataFrame in R/Python (like Pandas), but with rich optimizations under the hood.

**Key Characteristics:**
*   **Schema-on-read:** Unlike RDDs, DataFrames have a defined schema (column names and types), which allows Spark to optimize execution plans.
*   **Optimized:** Spark DataFrames utilize a powerful optimization engine called Catalyst Optimizer, which significantly improves performance.
*   **Distributed:** Like RDDs, DataFrames are distributed across a cluster, enabling parallel processing of large datasets.
*   **SQL-like operations:** You can perform operations using SQL queries or a rich set of DataFrame API functions.

**Why use DataFrames over RDDs?**
*   **Performance:** DataFrames generally offer better performance due to Catalyst Optimizer and Tungsten execution engine.
*   **Ease of Use:** The API is more intuitive and higher-level, resembling SQL or Pandas, making it easier to write complex data transformations.
*   **Readability:** Code written with DataFrames is often more concise and readable.

### PySpark DataFrame String Functions

PySpark provides a rich set of built-in functions to work with string data in DataFrames. These functions are part of the `pyspark.sql.functions` module and are optimized for performance.

Here are some commonly used string functions:

1.  **`concat(*cols)`**: Concatenates multiple string columns into a single string column.
    *   `df.select(concat(col("name"), lit(" - "), col("department")).alias("name_dept")).show()`

2.  **`length(col)`**: Returns the length of the string column.
    *   `df.select(col("name"), length(col("name")).alias("name_length")).show()`

3.  **`lower(col)`**: Converts the string column to lowercase.
    *   `df.select(col("name"), lower(col("name")).alias("lower_name")).show()`

4.  **`upper(col)`**: Converts the string column to uppercase.
    *   `df.select(col("name"), upper(col("name")).alias("upper_name")).show()`

5.  **`trim(col)`**: Removes leading and trailing whitespace from the string column.
    *   `df.select(col("city"), trim(col("city")).alias("trimmed_city")).show()`

6.  **`ltrim(col)`**: Removes leading whitespace from the string column.
    *   `df.select(col("city"), ltrim(col("city")).alias("ltrimmed_city")).show()`

7.  **`rtrim(col)`**: Removes trailing whitespace from the string column.
    *   `df.select(col("city"), rtrim(col("city")).alias("rtrimmed_city")).show()`

8.  **`substring(col, pos, len)`**: Extracts a substring from the string column. `pos` is 1-based index.
    *   `df.select(col("name"), substring(col("name"), 1, 3).alias("name_prefix")).show()`

9.  **`format_string(format, *cols)`**: Formats the string column using printf-style format strings.
    *   `df.select(format_string("Name: %s, Dept: %s", col("name"), col("department")).alias("formatted_info")).show()`

10. **`regexp_replace(col, pattern, replacement)`**: Replaces all substrings that match a Java regular expression `pattern` with `replacement`.
    *   `df.select(col("city"), regexp_replace(col("city"), "a", "*").alias("city_replaced")).show()`

**To use these functions, remember to import them from `pyspark.sql.functions`:**
